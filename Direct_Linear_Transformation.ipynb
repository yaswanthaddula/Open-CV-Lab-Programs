```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("Screenshot 2026-02-26 140552.png")

if img is None:
    print("Image not loaded")
else:
    img = cv2.resize(img, (800, 500))

    rows, cols = img.shape[:2]

    src_pts = np.float32([
        [100, 100],
        [cols-100, 100],
        [100, rows-100],
        [cols-100, rows-100]
    ])

    dst_pts = np.float32([
        [0, 0],
        [cols, 50],
        [50, rows],
        [cols-50, rows-50]
    ])

    H, status = cv2.findHomography(src_pts, dst_pts)

    result = cv2.warpPerspective(img, H, (cols, rows))

    print("Original Image")
    cv2_imshow(img)

    print("DLT Transformed Image")
    cv2_imshow(result)
```
