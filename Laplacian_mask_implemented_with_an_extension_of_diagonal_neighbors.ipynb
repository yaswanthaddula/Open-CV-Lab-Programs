```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/Screenshot 2026-03-02 194115.png')

# Define 8-neighbour Laplacian kernel
kernel = np.array([[1, 1, 1],
                   [1, -8, 1],
                   [1, 1, 1]])

# Apply filter
laplacian = cv2.filter2D(image, -1, kernel)

# True sharpening
sharpened = cv2.add(image, laplacian)

cv2_imshow(image)
cv2_imshow(sharpened)
```
