```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/Screenshot 2026-02-26 140552.png')

# Create kernel
kernel = np.ones((5,5), np.uint8)

# Apply erosion
eroded = cv2.erode(image, kernel, iterations=1)

cv2_imshow(image)
cv2_imshow(eroded)
```
