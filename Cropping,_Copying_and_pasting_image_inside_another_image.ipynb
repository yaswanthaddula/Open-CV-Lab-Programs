```python
import cv2
from google.colab.patches import cv2_imshow

# Read images
image1 = cv2.imread('/content/Screenshot 2026-03-02 194115.png')
image2 = cv2.imread('/content/Screenshot 2026-02-26 135502.png')

#Cropping
cropped = image1[100:300, 100:300]   # (y1:y2, x1:x2)

#Copying
copied = cropped.copy()
copied = cv2.resize(copied, (150, 150))

#Pasting inside another image
h, w = copied.shape[:2]
image2[50:50+h, 50:50+w] = copied

cv2_imshow(cropped)   # Cropped image
cv2_imshow(image2)    # Image after pasting
```
