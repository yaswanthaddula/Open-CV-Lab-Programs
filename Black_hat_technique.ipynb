```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/Screenshot 2026-03-02 194115.png')
kernel = np.ones((5,5), np.uint8)

# Apply Black Hat
blackhat = cv2.morphologyEx(image, cv2.MORPH_BLACKHAT, kernel)

cv2_imshow(image)
cv2_imshow(blackhat)
```
